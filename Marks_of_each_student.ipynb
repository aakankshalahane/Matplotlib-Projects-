```python
import matplotlib.pyplot as plt
import pandas as pd

# Creating a sample DataFrame for demonstration
data = {
    'Student': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    'Maths': [85, 90, 78, 92, 88],
    'Python': [70, 85, 90, 75, 95],
    'Database': [60, 75, 80, 85, 70]
}
df = pd.DataFrame(data)

plt.figure(figsize=(10, 5))

plt.plot(df["Student"], df["Maths"], marker="o", label="Maths")
plt.plot(df["Student"], df["Python"], marker="o", label="Python")
plt.plot(df["Student"], df["Database"], marker="o", label="Database")

plt.title("Student Performance by Subject")
plt.xlabel("Students")
plt.ylabel("Marks")

plt.xticks(rotation=45)
plt.legend()

plt.show()
```
