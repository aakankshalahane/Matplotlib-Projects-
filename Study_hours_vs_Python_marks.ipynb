```python
import pandas as pd
import numpy as np

# Sample data for demonstration
data = {
    'Study_Hours': np.random.rand(50) * 10, # Study hours from 0 to 10
    'Python': np.random.rand(50) * 100 # Python marks from 0 to 100
}
df = pd.DataFrame(data)

print("Sample DataFrame 'df' created:")
print(df.head())
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(df["Study_Hours"], df["Python"])

plt.title("Study Hours vs Python Marks")
plt.xlabel("Study Hours")
plt.ylabel("Python Marks")

plt.grid(True, alpha=0.3)

plt.show()
```
