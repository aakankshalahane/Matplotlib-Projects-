```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Sample data to define 'df', 'subjects', and 'average_marks'
data = {
    'Math': [70, 85, 60, 90, 75],
    'Science': [80, 92, 70, 85, 90],
    'English': [65, 78, 88, 70, 82],
    'History': [55, 68, 75, 80, 60]
}
df = pd.DataFrame(data)

subjects = list(df.columns)
average_marks = [df[subject].mean() for subject in subjects]

plt.figure(figsize=(9, 5))

x = range(len(subjects))

plt.bar(
    [i - 0.25 for i in x],
    average_marks,
    width=0.25,
    label="Average"
)

plt.bar(
    [i for i in x],
    [df[subject].max() for subject in subjects],
    width=0.25,
    label="Highest"
)

plt.bar(
    [i + 0.25 for i in x],
    [df[subject].min() for subject in subjects],
    width=0.25,
    label="Lowest"
)

plt.xticks(x, subjects)

plt.title("Subject Performance Comparison")
plt.xlabel("Subjects")
plt.ylabel("Marks")

plt.legend()

plt.show()
```
