```python
import matplotlib.pyplot as plt
import pandas as pd

# Placeholder data - Replace with your actual data
data = {
    'Student': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    'Math': [85, 90, 78, 92, 88],
    'Science': [70, 85, 90, 75, 80],
    'English': [90, 80, 85, 95, 75],
    'Python': [60, 75, 90, 80, 85],
    'Study_Hours': [5, 7, 6, 8, 4]
}
df = pd.DataFrame(data)
df['Average'] = df[['Math', 'Science', 'English', 'Python']].mean(axis=1)

subjects = ['Math', 'Science', 'English', 'Python']
average_marks = [df[subject].mean() for subject in subjects]

fig, axes = plt.subplots(2, 2, figsize=(14, 9))

# Average marks
axes[0, 0].bar(subjects, average_marks)
axes[0, 0].set_title("Average Marks")

# Python marks
axes[0, 1].hist(df["Python"], bins=5, edgecolor="black")
axes[0, 1].set_title("Python Marks Distribution")

# Study hours vs marks
axes[1, 0].scatter(df["Study_Hours"], df["Python"])
axes[1, 0].set_title("Study Hours vs Python Marks")
axes[1, 0].set_xlabel("Study Hours")
axes[1, 0].set_ylabel("Python Marks")

# Student averages
axes[1, 1].bar(df["Student"], df["Average"])
axes[1, 1].set_title("Student Average")
axes[1, 1].tick_params(axis="x", rotation=45)

plt.tight_layout()

plt.show()
```
