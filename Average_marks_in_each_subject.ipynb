```python
import pandas as pd
import matplotlib.pyplot as plt

# Sample DataFrame to demonstrate the fix
data = {
    'Student': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    'Maths': [85, 90, 78, 92, 88],
    'Python': [70, 85, 80, 90, 75],
    'Database': [65, 70, 75, 80, 82]
}
df = pd.DataFrame(data)

subjects = ["Maths", "Python", "Database"]

average_marks = [
    df["Maths"].mean(),
    df["Python"].mean(),
    df["Database"].mean()
]

plt.figure(figsize=(8, 5))

plt.bar(subjects, average_marks)

plt.title("Average Marks by Subject")
plt.xlabel("Subjects")
plt.ylabel("Average Marks")

plt.show()
```
